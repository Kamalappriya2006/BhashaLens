```python
import pandas as pd

train_df = pd.read_csv("/content/train.csv", encoding="utf-8")
val_df = pd.read_csv("/content/val.csv", encoding="utf-8")
test_df = pd.read_csv("/content/test.csv", encoding="utf-8")

print("TRAIN SHAPE:", train_df.shape)
print("VALIDATION SHAPE:", val_df.shape)
print("TEST SHAPE:", test_df.shape)

print("\nTotal samples:", len(train_df) + len(val_df) + len(test_df))
print("\nMissing values:")
print("Train:", train_df.isnull().sum().sum())
print("Validation:", val_df.isnull().sum().sum())
print("Test:", test_df.isnull().sum().sum())
```

```python
from datasets import Dataset

MAX_LENGTH = 512

def prepare_dataset(df):
    dataset = Dataset.from_pandas(
        df[["text", "label"]].copy(),
        preserve_index=False
    )

    def tokenize_function(examples):
        return tokenizer(
            examples["text"],
            truncation=True,
            max_length=MAX_LENGTH
        )

    tokenized = dataset.map(
        tokenize_function,
        batched=True,
        batch_size=500,
        remove_columns=["text"]
    )

    return tokenized

train_dataset = prepare_dataset(train_df)
val_dataset = prepare_dataset(val_df)
test_dataset = prepare_dataset(test_df)

print("Tokenization completed!")
print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
from torch.utils.data import DataLoader
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer,
    pad_to_multiple_of=8
)

columns = ["input_ids", "attention_mask", "label"]

train_dataset.set_format("torch", columns=columns)
val_dataset.set_format("torch", columns=columns)
test_dataset.set_format("torch", columns=columns)

train_loader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    collate_fn=data_collator
)

val_loader = DataLoader(
    val_dataset,
    batch_size=8,
    shuffle=False,
    collate_fn=data_collator
)

test_loader = DataLoader(
    test_dataset,
    batch_size=8,
    shuffle=False,
    collate_fn=data_collator
)

print("DataLoaders restored!")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))
```

```python
from datasets import Dataset

MAX_LENGTH = 512

def prepare_dataset(df):
    dataset = Dataset.from_pandas(
        df[["text", "label"]].copy(),
        preserve_index=False
    )

    def tokenize_function(examples):
        return tokenizer(
            examples["text"],
            truncation=True,
            max_length=MAX_LENGTH
        )

    tokenized = dataset.map(
        tokenize_function,
        batched=True,
        batch_size=500,
        remove_columns=["text"]
    )

    return tokenized

train_dataset = prepare_dataset(train_df)
val_dataset = prepare_dataset(val_df)
test_dataset = prepare_dataset(test_df)

print("TRAIN:", len(train_dataset))
print("VALIDATION:", len(val_dataset))
print("TEST:", len(test_dataset))

print("\nSample:")
print(train_dataset[0])
```

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2,
    id2label={
        0: "Human",
        1: "AI"
    },
    label2id={
        "Human": 0,
        "AI": 1
    }
)

print("XLM-RoBERTa model loaded successfully!")
print("Number of labels:", model.config.num_labels)
print("GPU:", torch.cuda.get_device_name(0))
print(
    "Total parameters:",
    sum(p.numel() for p in model.parameters())
)
```

```python
from torch.utils.data import DataLoader
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer,
    pad_to_multiple_of=8
)

columns = ["input_ids", "attention_mask", "label"]

train_dataset.set_format("torch", columns=columns)
val_dataset.set_format("torch", columns=columns)
test_dataset.set_format("torch", columns=columns)

train_loader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    collate_fn=data_collator
)

val_loader = DataLoader(
    val_dataset,
    batch_size=8,
    shuffle=False,
    collate_fn=data_collator
)

test_loader = DataLoader(
    test_dataset,
    batch_size=8,
    shuffle=False,
    collate_fn=data_collator
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))

batch = next(iter(train_loader))

print("\nInput shape:", batch["input_ids"].shape)
print("Attention mask shape:", batch["attention_mask"].shape)
print("Labels:", batch["labels"])
```

```python
import torch
import torch.nn as nn
from torch.optim import AdamW
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Calculate class weights
classes = np.array([0, 1])

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_df["label"].values
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to("cuda")

print("Class weights:")
print("Human (0):", class_weights[0].item())
print("AI (1):", class_weights[1].item())

# Optimizer
optimizer = AdamW(
    model.parameters(),
    lr=2e-5,
    weight_decay=0.01
)

# Weighted loss
criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

print("\nOptimizer configured!")
print("Learning rate:", 2e-5)
print("Weight decay:", 0.01)
print("Loss function: Weighted Cross Entropy")
```

```python
from transformers import get_linear_schedule_with_warmup
import math

EPOCHS = 3
GRADIENT_ACCUMULATION_STEPS = 4

total_optimizer_steps = math.ceil(
    len(train_loader) / GRADIENT_ACCUMULATION_STEPS
) * EPOCHS

warmup_steps = int(0.1 * total_optimizer_steps)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_optimizer_steps
)

scaler = torch.amp.GradScaler("cuda")

print("Epochs:", EPOCHS)
print("Gradient accumulation:", GRADIENT_ACCUMULATION_STEPS)
print("Total optimizer steps:", total_optimizer_steps)
print("Warmup steps:", warmup_steps)
print("Mixed precision: FP16")
print("Scheduler configured!")
```

```python
import torch
import numpy as np
from sklearn.metrics import accuracy_score, f1_score, classification_report
from tqdm.auto import tqdm

def train_one_epoch(model, train_loader):
    model.train()
    total_loss = 0
    optimizer.zero_grad(set_to_none=True)

    progress_bar = tqdm(train_loader, desc="Training")

    for step, batch in enumerate(progress_bar):
        input_ids = batch["input_ids"].to("cuda")
        attention_mask = batch["attention_mask"].to("cuda")
        labels = batch["labels"].to("cuda")

        with torch.amp.autocast("cuda", dtype=torch.float16):
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
            loss = criterion(outputs.logits, labels)
            loss = loss / GRADIENT_ACCUMULATION_STEPS

        scaler.scale(loss).backward()
        total_loss += loss.item() * GRADIENT_ACCUMULATION_STEPS

        if (
            (step + 1) % GRADIENT_ACCUMULATION_STEPS == 0
            or (step + 1) == len(train_loader)
        ):
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

            scaler.step(optimizer)
            scaler.update()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)

        progress_bar.set_postfix(
            loss=f"{total_loss / (step + 1):.4f}"
        )

    return total_loss / len(train_loader)


def evaluate_model(model, val_loader):
    model.eval()
    total_loss = 0
    all_predictions = []
    all_labels = []

    with torch.no_grad():
        for batch in tqdm(val_loader, desc="Validation"):
            input_ids = batch["input_ids"].to("cuda")
            attention_mask = batch["attention_mask"].to("cuda")
            labels = batch["labels"].to("cuda")

            with torch.amp.autocast("cuda", dtype=torch.float16):
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask
                )
                loss = criterion(outputs.logits, labels)

            total_loss += loss.item()

            predictions = torch.argmax(outputs.logits, dim=1)

            all_predictions.extend(predictions.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    accuracy = accuracy_score(all_labels, all_predictions)
    macro_f1 = f1_score(
        all_labels, all_predictions, average="macro"
    )

    return {
        "val_loss": total_loss / len(val_loader),
        "val_accuracy": accuracy,
        "val_macro_f1": macro_f1,
        "predictions": all_predictions,
        "labels": all_labels
    }


print("Training and evaluation functions created successfully!")
```

```python
import json
import os

training_history = []
best_macro_f1 = 0.0

for epoch in range(EPOCHS):

    print(f"\n{'='*50}")
    print(f"EPOCH {epoch + 1}/{EPOCHS}")
    print(f"{'='*50}")

    train_loss = train_one_epoch(model, train_loader)

    val_results = evaluate_model(model, val_loader)

    epoch_results = {
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "val_loss": val_results["val_loss"],
        "val_accuracy": val_results["val_accuracy"],
        "val_macro_f1": val_results["val_macro_f1"]
    }

    training_history.append(epoch_results)

    print(f"\nEpoch {epoch + 1} Results")
    print(f"Training Loss: {train_loss:.4f}")
    print(f"Validation Loss: {val_results['val_loss']:.4f}")
    print(f"Validation Accuracy: {val_results['val_accuracy']:.4f}")
    print(f"Validation Macro F1: {val_results['val_macro_f1']:.4f}")

    # Save epoch checkpoint
    epoch_dir = os.path.join(
        CHECKPOINT_DIR, f"epoch_{epoch + 1}"
    )
    os.makedirs(epoch_dir, exist_ok=True)

    model.save_pretrained(epoch_dir)
    tokenizer.save_pretrained(epoch_dir)

    # Save training history
    with open(
        os.path.join(CHECKPOINT_DIR, "training_history.json"), "w"
    ) as f:
        json.dump(training_history, f, indent=4)

    # Save best model
    if val_results["val_macro_f1"] > best_macro_f1:
        best_macro_f1 = val_results["val_macro_f1"]

        best_dir = os.path.join(CHECKPOINT_DIR, "best_model")
        os.makedirs(best_dir, exist_ok=True)

        model.save_pretrained(best_dir)
        tokenizer.save_pretrained(best_dir)

        print("Best model updated!")

    print(f"Checkpoint saved: {epoch_dir}")

print("\nTRAINING COMPLETED!")
print("Best Validation Macro F1:", best_macro_f1)
```
